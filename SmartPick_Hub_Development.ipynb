```python
js_code = """
// SMARTPICK HUB - JAVASCRIPT

document.addEventListener("DOMContentLoaded", function () {

    console.log("SmartPick Hub website loaded successfully!");

    // Display website current year in footer
    const footer = document.querySelector(".footer");

    if (footer) {
        const paragraphs = footer.querySelectorAll("p");
        const yearText = paragraphs[paragraphs.length - 1];

        if (yearText) {
            yearText.textContent =
                "© " + new Date().getFullYear() +
                " SmartPick Hub. All rights reserved.";
        }
    }

});
"""

js_path = project / "script.js"
js_path.write_text(js_code, encoding="utf-8")

print("SUCCESS! script.js created.")
print("File location:", js_path)
```

```python
from pathlib import Path

project = Path("/content/SmartPick-Hub")

print("SMARTPICK HUB PROJECT FILES")
print("---------------------------")

for filename in ["index.html", "style.css", "script.js"]:
    file = project / filename

    if file.exists():
        print("PASS -", filename, "-", file.stat().st_size, "bytes")
    else:
        print("FAIL -", filename)

print("---------------------------")
print("Project structure:")
for path in sorted(project.rglob("*")):
    print(path.relative_to(project))
```

```python
import subprocess
import time
from google.colab import output

# Start local website server
server = subprocess.Popen(
    [
        "python", "-m", "http.server", "8000",
        "--directory", "/content/SmartPick-Hub"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(2)

print("SmartPick Hub server started!")
print("Opening website preview...")

output.serve_kernel_port_as_iframe(8000, height=750)
```

```python
import subprocess
import time
import socket
from google.colab import output

PROJECT = "/content/SmartPick-Hub"

def port_is_open(port):
    sock = socket.socket()
    result = sock.connect_ex(("127.0.0.1", port))
    sock.close()
    return result == 0

# Check whether the previous server is still running
if "server" in globals() and server.poll() is None:
    port = 8000
    print("Existing website server is running.")

elif port_is_open(8000):
    port = 8000
    print("Existing server detected on port 8000.")

else:
    port = 8000
    server = subprocess.Popen(
        [
            "python", "-m", "http.server", str(port),
            "--directory", PROJECT
        ],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )
    time.sleep(2)

    if not port_is_open(port):
        raise RuntimeError("Website server could not start.")

    print("New website server started.")

print("Opening SmartPick Hub preview...")
output.serve_kernel_port_as_iframe(port, height=800)
```

```python
from pathlib import Path
import re

html = Path("/content/SmartPick-Hub/index.html").read_text(
    encoding="utf-8"
)

classes = sorted(set(re.findall(r'class=["\']([^"\']+)["\']', html)))
ids = sorted(set(re.findall(r'id=["\']([^"\']+)["\']', html)))

print("========== HTML DESIGN STRUCTURE ==========")

print("\nCSS CLASSES:")
for item in classes:
    print("-", item)

print("\nHTML SECTION IDs:")
for item in ids:
    print("-", item)

print("\n========== END ==========")
```

```python
from pathlib import Path
import requests

folder = Path("/content/SmartPick-Hub")

print("===== LOCAL FILE CHECK =====")

for filename in ["index.html", "style.css", "script.js"]:
    path = folder / filename
    print(f"\n{filename}:")
    print("Exists:", path.exists())
    if path.exists():
        print("Size:", path.stat().st_size, "bytes")

html = (folder / "index.html").read_text(encoding="utf-8")

print("\n===== LOCAL DESIGN CHECK =====")
print("New hero:", "Choose Smarter." in html)
print("Old hero:", "Find Better Products." in html)
print("New logo:", "logo-mark" in html)

print("\n===== SERVER OUTPUT CHECK =====")

try:
    response = requests.get("http://localhost:8000/index.html")
    served_html = response.text

    print("Server status:", response.status_code)
    print("New hero served:", "Choose Smarter." in served_html)
    print("Old hero served:", "Find Better Products." in served_html)

except Exception as e:
    print("Server error:", e)
```

```python
import os
import re
import time
import subprocess
import urllib.request

# 1. Download Cloudflare Tunnel
download = subprocess.run([
    "wget", "-q",
    "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64",
    "-O", "/content/cloudflared"
])

if download.returncode != 0:
    print("Download failed. Please run the cell again.")
else:
    os.chmod("/content/cloudflared", 0o755)

    # 2. Check website server
    try:
        response = urllib.request.urlopen(
            "http://127.0.0.1:8001/index.html"
        )
        print("Website server:", response.status)

        # 3. Start public tunnel
        log_file = open("/content/cloudflared.log", "w")

        tunnel = subprocess.Popen(
            [
                "/content/cloudflared",
                "tunnel",
                "--url", "http://127.0.0.1:8001",
                "--no-autoupdate"
            ],
            stdout=log_file,
            stderr=subprocess.STDOUT
        )

        # 4. Find public URL
        public_url = None

        for i in range(30):
            time.sleep(2)

            with open("/content/cloudflared.log", "r") as f:
                logs = f.read()

            match = re.search(
                r"https://[a-zA-Z0-9-]+\.trycloudflare\.com",
                logs
            )

            if match:
                public_url = match.group(0)
                break

        if public_url:
            print("\nSMARTPICK HUB PUBLIC MOBILE URL:")
            print(public_url + "/index.html")
        else:
            print("Public URL generate nahi hua. Latest logs:")
            print(logs[-2000:])

    except Exception as e:
        print("Error:", e)
```

```python
# STEP 1: SmartPick Hub SEO Audit
import os
from pathlib import Path
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub")

if not PROJECT.exists():
    raise FileNotFoundError(
        "Project folder nahi mila. Check karein: /content/SmartPick-Hub"
    )

print("PROJECT:", PROJECT)
print("\n========== FILE STRUCTURE ==========")

for item in sorted(PROJECT.iterdir()):
    if item.is_file():
        print("FILE  :", item.name)
    elif item.is_dir():
        print("FOLDER:", item.name)

# Homepage check
html_file = PROJECT / "index.html"

if not html_file.exists():
    raise FileNotFoundError("index.html project folder mein nahi mili.")

html = html_file.read_text(encoding="utf-8")
soup = BeautifulSoup(html, "html.parser")

print("\n========== HOMEPAGE SEO AUDIT ==========")

title = soup.title.get_text(strip=True) if soup.title else "MISSING"
description = soup.find("meta", attrs={"name": "description"})
h1_tags = soup.find_all("h1")
canonical = soup.find("link", rel="canonical")

print("Page Title       :", title)
print("Meta Description :", description.get("content", "MISSING") if description else "MISSING")
print("H1 Count         :", len(h1_tags))
print("H1 Text          :", [h.get_text(" ", strip=True) for h in h1_tags])
print("Canonical URL    :", canonical.get("href", "MISSING") if canonical else "MISSING")

print("\n========== SEO FILES ==========")
for filename in ["robots.txt", "sitemap.xml"]:
    path = PROJECT / filename
    print(filename, ":", "EXISTS" if path.exists() else "NOT CREATED")

print("\n========== CONNECTED ASSETS ==========")
print("CSS files:", [x.get("href") for x in soup.find_all("link", href=True) if x.get("rel") and "stylesheet" in x.get("rel")])
print("JS files :", [x.get("src") for x in soup.find_all("script", src=True)])

print("\n========== AUDIT COMPLETE ==========")
```

```python
import os
import shutil
import zipfile
from datetime import datetime

PROJECT = "/content/SmartPick-Hub"

# Backup filename with date and time
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
BACKUP = f"/content/SmartPick-Hub-Backup-{timestamp}.zip"

print("SMARTPICK HUB - BACKUP")
print("-" * 40)

if not os.path.exists(PROJECT):
    print("FAIL - Project folder not found")

else:
    # Create ZIP backup
    shutil.make_archive(
        BACKUP.replace(".zip", ""),
        "zip",
        root_dir="/content",
        base_dir="SmartPick-Hub"
    )

    # Verify ZIP
    with zipfile.ZipFile(BACKUP, "r") as zip_file:
        result = zip_file.testzip()

    if result is None:
        print("PASS - ZIP backup created")
        print("PASS - ZIP integrity verified")
        print("Backup location:", BACKUP)
        print("Backup size:", round(os.path.getsize(BACKUP) / 1024, 2), "KB")
    else:
        print("FAIL - ZIP verification error:", result)
```

```python
from pathlib import Path

path = Path("/content/SmartPick-Hub/script.js")

new_js = '''document.addEventListener("DOMContentLoaded", function () {

  console.log("SmartPick Hub loaded successfully!");

  // Automatic copyright year
  const copyright = document.querySelector(".footer-bottom p:first-child");

  if (copyright) {
    copyright.textContent = "© " + new Date().getFullYear() +
      " SmartPick Hub. All rights reserved.";
  }

  // Mobile navigation
  const menuToggle = document.getElementById("menu-toggle");
  const siteNav = document.getElementById("site-nav");

  if (menuToggle && siteNav) {

    menuToggle.addEventListener("click", function () {

      const isOpen = siteNav.classList.toggle("menu-open");

      menuToggle.setAttribute("aria-expanded", String(isOpen));

      menuToggle.setAttribute(
        "aria-label",
        isOpen ? "Close navigation menu" : "Open navigation menu"
      );

      const menuIcon = menuToggle.querySelector(".menu-icon");

      if (menuIcon) {
        menuIcon.textContent = isOpen ? "✕" : "☰";
      }

    });

    // Close menu after selecting a navigation link
    siteNav.querySelectorAll("a").forEach(function (link) {

      link.addEventListener("click", function () {

        siteNav.classList.remove("menu-open");

        menuToggle.setAttribute("aria-expanded", "false");
        menuToggle.setAttribute("aria-label", "Open navigation menu");

        const menuIcon = menuToggle.querySelector(".menu-icon");

        if (menuIcon) {
          menuIcon.textContent = "☰";
        }

      });

    });

  }

});
'''

path.write_text(new_js, encoding="utf-8")

saved_js = path.read_text(encoding="utf-8")

checks = {
    "Copyright functionality preserved": "copyright.textContent" in saved_js,
    "Menu button selected": 'getElementById("menu-toggle")' in saved_js,
    "Navigation selected": 'getElementById("site-nav")' in saved_js,
    "Menu open/close function": 'classList.toggle("menu-open")' in saved_js,
    "Accessibility state updated": 'aria-expanded' in saved_js,
    "Menu closes after link click": 'siteNav.querySelectorAll("a")' in saved_js
}

print("JAVASCRIPT VERIFICATION")
print("-" * 40)

for name, result in checks.items():
    print(("PASS" if result else "FAIL"), "-", name)

print("\nTotal:", sum(checks.values()), "/", len(checks))
```

```python
import subprocess
import time
import re
from pathlib import Path

# Check existing website server
if "mobile_server" not in globals() or mobile_server.poll() is not None:
    raise RuntimeError("Website server is not running. Pehle website server start karein.")

# Install Cloudflare Tunnel
binary = "/usr/local/bin/cloudflared"

if not Path(binary).exists():
    print("Installing Cloudflare Tunnel...")

    subprocess.run([
        "wget", "-q",
        "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64",
        "-O", binary
    ], check=True)

    subprocess.run(["chmod", "+x", binary], check=True)

print("PASS - Cloudflare Tunnel installed")

# Start public tunnel to your existing website server
log_file = open("/tmp/smartpick-tunnel.log", "w")

tunnel = subprocess.Popen(
    [
        binary,
        "tunnel",
        "--url", f"http://127.0.0.1:{PORT}",
        "--no-autoupdate"
    ],
    stdout=log_file,
    stderr=subprocess.STDOUT
)

# Wait for public URL
public_url = None

for i in range(30):
    time.sleep(2)

    log_file.flush()
    logs = Path("/tmp/smartpick-tunnel.log").read_text(
        errors="ignore"
    )

    match = re.search(
        r"https://[a-zA-Z0-9-]+\.trycloudflare\.com",
        logs
    )

    if match:
        public_url = match.group(0)
        break

if public_url:
    print("\nPASS - Public tunnel created")
    print("\nSMARTPICK HUB MOBILE URL:")
    print(public_url)
else:
    print("Tunnel URL not found. Last logs:")
    print(logs[-2500:])
```

```python
from pathlib import Path
from bs4 import BeautifulSoup
from copy import deepcopy

project = Path("/content/SmartPick-Hub")
source = project / "index.html"

# Read original website
original = BeautifulSoup(
    source.read_text(encoding="utf-8"),
    "html.parser"
)

# Page configuration
pages = {
    "index.html": {
        "title": "SmartPick Hub | Smart Choices. Better Living.",
        "description": "Discover useful technology, AI tools and practical recommendations for smarter everyday decisions.",
        "sections": ["hero"]
    },
    "technology.html": {
        "title": "Technology & AI Tools | SmartPick Hub",
        "description": "Explore AI software, technology, gadgets and useful digital tools.",
        "sections": ["technology"]
    },
    "featured.html": {
        "title": "Featured Topics | SmartPick Hub",
        "description": "Explore featured AI writing, AI design and productivity software topics.",
        "sections": ["featured"]
    },
    "guides.html": {
        "title": "Buying Guides | SmartPick Hub",
        "description": "Explore practical buying guides, product research and comparisons.",
        "sections": ["guides"]
    },
    "about.html": {
        "title": "About Us | SmartPick Hub",
        "description": "Learn about SmartPick Hub, our mission and our commitment to helpful information.",
        "sections": ["about"]
    },
    "affiliate-disclosure.html": {
        "title": "Affiliate Disclosure | SmartPick Hub",
        "description": "Read how affiliate links and commissions work on SmartPick Hub.",
        "sections": ["disclosure"]
    }
}

# Navigation destinations
nav_links = {
    "Home": "index.html",
    "Technology": "technology.html",
    "Featured": "featured.html",
    "Buying Guides": "guides.html",
    "About": "about.html",
    "Explore Picks →": "featured.html"
}

# Generate individual pages
for filename, config in pages.items():

    page = BeautifulSoup(str(original), "html.parser")

    # Set page title
    page.title.string = config["title"]

    # Update meta description
    description = page.find("meta", attrs={"name": "description"})
    if description:
        description["content"] = config["description"]

    # Update header navigation
    nav = page.find("nav", id="site-nav")

    if nav:
        for link in nav.find_all("a"):
            label = link.get_text(" ", strip=True)

            if label in nav_links:
                link["href"] = nav_links[label]

            if nav_links.get(label) == filename:
                link["aria-current"] = "page"
            else:
                link.attrs.pop("aria-current", None)

    # Update footer links
    footer = page.find("footer")

    if footer:
        for link in footer.select(".footer-links a"):
            label = link.get_text(" ", strip=True)

            if label in nav_links:
                link["href"] = nav_links[label]

    # Create new main content
    old_main = page.find("main")
    new_main = page.new_tag("main")

    for section_id in config["sections"]:
        section = original.find(id=section_id)

        if section:
            new_main.append(deepcopy(section))

    old_main.replace_with(new_main)

    # Update internal links in content
    content_links = {
        "#technology": "technology.html",
        "#featured": "featured.html",
        "#guides": "guides.html",
        "#about": "about.html"
    }

    for link in new_main.find_all("a"):
        href = link.get("href")

        if href in content_links:
            link["href"] = content_links[href]

    # Save page
    output = project / filename
    output.write_text(
        "<!DOCTYPE html>\n" + str(page),
        encoding="utf-8"
    )

    print("PASS - Created:", filename)

print("\nMULTI-PAGE GENERATION COMPLETE")
print("Total pages:", len(pages))
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

project = Path("/content/SmartPick-Hub")

expected_pages = [
    "index.html",
    "technology.html",
    "featured.html",
    "guides.html",
    "about.html",
    "affiliate-disclosure.html"
]

print("SMARTPICK HUB MULTI-PAGE AUDIT")
print("-" * 45)

passed = 0

for filename in expected_pages:

    path = project / filename

    if not path.exists():
        print("FAIL -", filename, "missing")
        continue

    soup = BeautifulSoup(
        path.read_text(encoding="utf-8"),
        "html.parser"
    )

    checks = {
        "HTML title": bool(soup.title and soup.title.string),
        "Header": soup.find("header") is not None,
        "Navigation": soup.find("nav", id="site-nav") is not None,
        "Main content": soup.find("main") is not None,
        "Footer": soup.find("footer") is not None,
        "CSS connected": soup.find("link", href="style.css") is not None,
        "JavaScript connected": soup.find("script", src="script.js") is not None
    }

    result = all(checks.values())

    print(
        ("PASS" if result else "FAIL"),
        "-", filename,
        "|",
        sum(checks.values()), "/ 7"
    )

    if result:
        passed += 1

print("\nFINAL RESULT:", passed, "/", len(expected_pages), "pages passed")
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

project = Path("/content/SmartPick-Hub")

page_sections = {
    "index.html": ["hero"],
    "technology.html": ["technology"],
    "featured.html": ["featured"],
    "guides.html": ["guides"],
    "about.html": ["about"],
    "affiliate-disclosure.html": ["disclosure"]
}

expected_navigation = {
    "Home": "index.html",
    "Technology": "technology.html",
    "Featured": "featured.html",
    "Buying Guides": "guides.html",
    "About": "about.html"
}

print("SMARTPICK HUB PAGE SEPARATION TEST")
print("-" * 50)

total_pass = 0
total_checks = 0

for filename, expected_ids in page_sections.items():

    soup = BeautifulSoup(
        (project / filename).read_text(encoding="utf-8"),
        "html.parser"
    )

    actual_ids = [
        section.get("id")
        for section in soup.select("main section")
    ]

    content_pass = actual_ids == expected_ids

    nav = soup.find("nav", id="site-nav")
    nav_pass = True

    for label, destination in expected_navigation.items():
        link = next(
            (a for a in nav.find_all("a")
             if a.get_text(" ", strip=True) == label),
            None
        )

        if not link or link.get("href") != destination:
            nav_pass = False

    print(f"\n{filename}")
    print("Content separation:", "PASS" if content_pass else "FAIL")
    print("Navigation links:", "PASS" if nav_pass else "FAIL")
    print("Page sections:", actual_ids)

    total_pass += int(content_pass) + int(nav_pass)
    total_checks += 2

print("\nFINAL RESULT:", total_pass, "/", total_checks)
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub")

# ---------------------------------
# FIX 1: HOMEPAGE
# ---------------------------------

home_file = PROJECT / "index.html"

soup = BeautifulSoup(
    home_file.read_text(encoding="utf-8"),
    "html.parser"
)

hero = soup.select_one("main section.hero")

if hero:
    hero["id"] = "home"
    home_file.write_text(str(soup), encoding="utf-8")
    print("PASS - Homepage section ID added")
else:
    print("FAIL - Homepage hero section not found")


# ---------------------------------
# FIX 2: AFFILIATE DISCLOSURE
# ---------------------------------

disclosure_file = PROJECT / "affiliate-disclosure.html"

soup = BeautifulSoup(
    disclosure_file.read_text(encoding="utf-8"),
    "html.parser"
)

disclosure_section = None

for section in soup.select("main section"):
    heading = section.find(["h1", "h2", "h3"])

    if heading and "affiliate disclosure" in heading.get_text(
        " ", strip=True
    ).lower():
        disclosure_section = section
        break

if disclosure_section:
    disclosure_section["id"] = "affiliate-disclosure"

    disclosure_file.write_text(
        str(soup),
        encoding="utf-8"
    )

    print("PASS - Affiliate Disclosure section ID added")
else:
    print("FAIL - Affiliate Disclosure section not found")
```

```python
from pathlib import Path
from bs4 import BeautifulSoup
from copy import deepcopy
import shutil

PROJECT = Path("/content/SmartPick-Hub")

HOME = PROJECT / "index.html"
DISCLOSURE = PROJECT / "affiliate-disclosure.html"

# Read restored homepage
home_soup = BeautifulSoup(
    HOME.read_text(encoding="utf-8"),
    "html.parser"
)

# Find original Affiliate Disclosure section
source_section = None

for element in home_soup.main.find_all(
    ["section", "div"],
    recursive=True
):
    heading = element.find(["h1", "h2", "h3"])

    if heading and "affiliate disclosure" in heading.get_text(
        " ", strip=True
    ).lower():
        source_section = element
        break

if source_section is None:
    print("FAIL - Original disclosure content not found")

else:
    # Safety copy
    shutil.copy2(
        DISCLOSURE,
        "/content/affiliate-disclosure-before-recovery.html"
    )

    # Read existing disclosure page shell
    page_soup = BeautifulSoup(
        DISCLOSURE.read_text(encoding="utf-8"),
        "html.parser"
    )

    # Preserve header, footer and navigation
    main = page_soup.find("main")

    if main is None:
        print("FAIL - Main element not found in disclosure page")

    else:
        main.clear()

        restored_section = deepcopy(source_section)
        restored_section["id"] = "affiliate-disclosure"

        main.append(restored_section)

        # Update page title
        if page_soup.title:
            page_soup.title.string = (
                "Affiliate Disclosure | SmartPick Hub"
            )

        DISCLOSURE.write_text(
            str(page_soup),
            encoding="utf-8"
        )

        # Verify
        check = BeautifulSoup(
            DISCLOSURE.read_text(encoding="utf-8"),
            "html.parser"
        )

        print("PASS - Affiliate Disclosure content restored")
        print(
            "PASS - Section ID:",
            check.main.find(id="affiliate-disclosure") is not None
        )
        print(
            "PASS - Content characters:",
            len(check.main.get_text(" ", strip=True))
        )
        print(
            "PASS - Header preserved:",
            check.header is not None
        )
        print(
            "PASS - Footer preserved:",
            check.footer is not None
        )
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub")

pages = {
    "index.html": ["home"],
    "technology.html": ["technology"],
    "featured.html": ["featured"],
    "guides.html": ["guides"],
    "about.html": ["about"],
    "affiliate-disclosure.html": ["affiliate-disclosure"]
}

print("SMARTPICK HUB FINAL PAGE VERIFICATION")
print("-" * 55)

total_pass = 0
total_tests = 0

for filename, expected_ids in pages.items():

    path = PROJECT / filename

    print(f"\n{filename}")

    if not path.exists():
        print("FAIL - File missing")
        total_tests += 2
        continue

    soup = BeautifulSoup(
        path.read_text(encoding="utf-8"),
        "html.parser"
    )

    # Test 1: Main content and expected section
    main = soup.find("main")

    content_pass = (
        main is not None
        and bool(main.get_text(" ", strip=True))
        and all(main.find(id=section_id) for section_id in expected_ids)
    )

    print(
        "Content separation:",
        "PASS" if content_pass else "FAIL"
    )

    # Test 2: Navigation links
    nav = soup.find("nav")

    nav_pass = (
        nav is not None
        and len(nav.find_all("a", href=True)) >= 5
    )

    print(
        "Navigation links:",
        "PASS" if nav_pass else "FAIL"
    )

    print(
        "Page sections:",
        [
            section_id
            for section_id in expected_ids
            if main and main.find(id=section_id)
        ]
    )

    total_pass += int(content_pass) + int(nav_pass)
    total_tests += 2

print("\n" + "-" * 55)
print(f"FINAL RESULT: {total_pass} / {total_tests}")
```

```python
from pathlib import Path
from datetime import datetime
import zipfile
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub")

# 1. Check project
if not PROJECT.exists():
    raise FileNotFoundError("SmartPick-Hub folder not found!")

# 2. Create complete backup
backup_name = (
    "SmartPick-Hub-12-12-PASS-"
    + datetime.now().strftime("%Y%m%d_%H%M%S")
    + ".zip"
)

BACKUP = Path("/content") / backup_name

with zipfile.ZipFile(
    BACKUP, "w", zipfile.ZIP_DEFLATED
) as z:

    for file in PROJECT.rglob("*"):
        if file.is_file():
            z.write(file, file.relative_to(PROJECT.parent))

print("PASS - Complete website backup created")
print("BACKUP:", BACKUP)
print("SIZE:", round(BACKUP.stat().st_size / 1024, 2), "KB")

# 3. Audit all six pages
pages = [
    "index.html",
    "technology.html",
    "featured.html",
    "guides.html",
    "about.html",
    "affiliate-disclosure.html"
]

print("\nSMARTPICK HUB PAGE CONTENT AUDIT")
print("=" * 55)

for filename in pages:

    path = PROJECT / filename

    print("\nPAGE:", filename)

    if not path.exists():
        print("FAIL - File missing")
        continue

    soup = BeautifulSoup(
        path.read_text(encoding="utf-8"),
        "html.parser"
    )

    main = soup.find("main")

    if not main:
        print("FAIL - Main content missing")
        continue

    print("Title:", soup.title.get_text(strip=True) if soup.title else "Missing")
    print("Content characters:", len(main.get_text(" ", strip=True)))

    print("Sections:")

    for section in main.find_all(
        ["section", "article"],
        recursive=False
    ):
        headings = section.find_all(
            ["h1", "h2", "h3"],
            recursive=True
        )

        heading_text = (
            headings[0].get_text(" ", strip=True)
            if headings else "No heading"
        )

        print(
            " -",
            heading_text,
            "| ID:", section.get("id"),
            "| CLASS:", section.get("class")
        )

    print("Links:")

    for link in main.find_all("a", href=True):
        print(
            " -",
            link.get_text(" ", strip=True),
            "=>",
            link["href"]
        )

print("\nAUDIT COMPLETE")
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub")
HOME = PROJECT / "index.html"

soup = BeautifulSoup(
    HOME.read_text(encoding="utf-8"),
    "html.parser"
)

main = soup.find("main")

if not main:
    raise Exception("Homepage main section missing!")

# 1. Hero buttons
hero = main.find("section", id="home")

if hero:
    for link in hero.find_all("a", href=True):
        text = link.get_text(" ", strip=True).lower()

        if "technology" in text:
            link["href"] = "technology.html"

        elif "buying guides" in text:
            link["href"] = "guides.html"

# 2. Technology preview links
technology = main.find("section", id="technology")

if technology:
    for link in technology.find_all("a", href=True):
        link["href"] = "technology.html"

# 3. Featured preview links
featured = main.find("section", id="featured")

if featured:
    for link in featured.find_all("a", href=True):
        link["href"] = "featured.html"

# 4. Add a dedicated page CTA if missing
def add_section_cta(section, href, label):

    if section is None:
        return

    existing = section.find(
        "a",
        href=href
    )

    if existing:
        return

    wrapper = soup.new_tag("div")
    wrapper["class"] = ["section-cta"]

    link = soup.new_tag(
        "a",
        href=href
    )

    link["class"] = ["btn", "btn-outline"]
    link.string = label

    wrapper.append(link)
    section.append(wrapper)

# Technology and Featured CTAs
add_section_cta(
    technology,
    "technology.html",
    "Explore All Technology →"
)

add_section_cta(
    featured,
    "featured.html",
    "Explore All Featured Topics →"
)

# 5. Buying Guides preview CTA
guides = main.find("section", id="guides")

add_section_cta(
    guides,
    "guides.html",
    "Explore All Buying Guides →"
)

# 6. About preview CTA
about = main.find("section", id="about")

add_section_cta(
    about,
    "about.html",
    "Discover Our Story →"
)

# 7. Affiliate disclosure notice link
disclosure = main.find(
    "section",
    class_="disclosure"
)

add_section_cta(
    disclosure,
    "affiliate-disclosure.html",
    "Read Affiliate Disclosure →"
)

# 8. Save homepage
HOME.write_text(
    str(soup),
    encoding="utf-8"
)

print("SMARTPICK HUB HOMEPAGE UPGRADE")
print("-" * 45)

print("PASS - Hero navigation updated")
print("PASS - Technology preview links updated")
print("PASS - Featured preview links updated")
print("PASS - Buying Guides CTA added")
print("PASS - About CTA added")
print("PASS - Affiliate Disclosure CTA added")
print("PASS - Original homepage sections preserved")
print("PASS - Homepage saved")
```

```python
from pathlib import Path
import re

CSS = Path("/content/SmartPick-Hub/style.css")

css = CSS.read_text(encoding="utf-8")

print("SMARTPICK HUB CSS DESIGN AUDIT")
print("=" * 50)

selectors = [
    "body",
    ".hero",
    ".hero-content",
    ".hero-visual",
    ".section",
    ".section-heading",
    ".card-grid",
    ".category-card",
    ".topic-card",
    ".guides",
    ".about",
    ".disclosure",
    ".btn",
    ".footer"
]

for selector in selectors:
    found = re.search(
        re.escape(selector) + r"\s*\{",
        css
    )

    print(
        ("PASS" if found else "MISSING"),
        "-",
        selector
    )

print("\nRESPONSIVE BREAKPOINTS:")

breakpoints = re.findall(
    r"@media[^{]+",
    css
)

for breakpoint in breakpoints:
    print("-", breakpoint.strip())

print("\nCSS SIZE:", len(css), "characters")
print("CSS AUDIT COMPLETE")
```

```python
from pathlib import Path
from bs4 import BeautifulSoup
import re

BASE = Path("/content/SmartPick-Hub")

html = (BASE / "index.html").read_text(encoding="utf-8")
css = (BASE / "style.css").read_text(encoding="utf-8")

soup = BeautifulSoup(html, "html.parser")

print("SMARTPICK HUB HTML + CSS MATCHING AUDIT")
print("=" * 55)

print("\n1. HOMEPAGE SECTION STRUCTURE")

main = soup.find("main")

if main:
    for i, section in enumerate(main.find_all("section", recursive=False), 1):
        heading = section.find(["h1", "h2", "h3"])
        title = heading.get_text(" ", strip=True) if heading else "No heading"

        print(f"\nSection {i}: {title}")
        print("ID:", section.get("id", "None"))
        print("Classes:", section.get("class", []))

        cards = section.select("article, .card, .topic-card")
        print("Card elements:", len(cards))

print("\n2. ALL HOMEPAGE CSS CLASSES")

classes = sorted({
    class_name
    for element in soup.find_all(class_=True)
    for class_name in element.get("class", [])
})

for name in classes:
    exists = bool(re.search(
        r"\." + re.escape(name) + r"(?![\w-])",
        css
    ))
    print(("CSS FOUND" if exists else "NO CSS RULE"), "-", name)

print("\n3. EXISTING CSS DESIGN SELECTORS")

for selector in [
    ".hero", ".hero-visual", ".section",
    ".section-heading", ".card-grid",
    ".topic-card", ".about", ".disclosure",
    ".btn", ".footer", ".site-header",
    ".site-nav", ".menu-toggle"
]:
    found = bool(re.search(
        re.escape(selector) + r"(?![\w-])",
        css
    ))
    print(("FOUND" if found else "NOT FOUND"), "-", selector)

print("\n4. FILE INFORMATION")
print("Homepage HTML:", len(html), "characters")
print("Stylesheet CSS:", len(css), "characters")

print("\nAUDIT COMPLETE")
```

```python
from pathlib import Path
import re

BASE = Path("/content/SmartPick-Hub")
css = (BASE / "style.css").read_text(encoding="utf-8")

print("SMARTPICK HUB CSS INTEGRITY TEST")
print("=" * 50)

checks = {
    "Redesign marker exists":
        "/* SMARTPICK HUB PREMIUM HOMEPAGE REDESIGN */" in css,

    "Hero styling exists":
        ".hero {" in css,

    "Category cards styling exists":
        ".category-card," in css,

    "Featured cards styling exists":
        ".topic-card {" in css,

    "Buying guides styling exists":
        ".guide-placeholder {" in css,

    "About styling exists":
        ".about {" in css,

    "Disclosure styling exists":
        ".disclosure {" in css,

    "Tablet responsive breakpoint":
        "@media (max-width: 900px)" in css,

    "Mobile responsive breakpoint":
        "@media (max-width: 650px)" in css,

    "Accessibility focus support":
        ":focus-visible" in css,

    "Reduced motion support":
        "prefers-reduced-motion" in css
}

for name, passed in checks.items():
    print(("PASS" if passed else "FAIL"), "-", name)

print("\nCSS BRACES CHECK")

opens = css.count("{")
closes = css.count("}")

print("Opening braces:", opens)
print("Closing braces:", closes)

brace_pass = opens == closes
print(("PASS" if brace_pass else "FAIL"), "- CSS brace balance")

print("\nTOTAL:", sum(checks.values()) + brace_pass, "/", len(checks) + 1)
print("CSS SIZE:", len(css), "characters")
```

```python
from pathlib import Path
import http.server
import socketserver
import threading
import time

BASE = Path("/content/SmartPick-Hub")
PORT = 8000

if not BASE.exists():
    print("ERROR - SmartPick-Hub folder not found")
else:
    class ReusableTCPServer(socketserver.TCPServer):
        allow_reuse_address = True

    try:
        server = ReusableTCPServer(
            ("", PORT),
            lambda *args, **kwargs: http.server.SimpleHTTPRequestHandler(
                *args,
                directory=str(BASE),
                **kwargs
            )
        )

        thread = threading.Thread(
            target=server.serve_forever,
            daemon=True
        )
        thread.start()

        time.sleep(1)

        print("PASS - SmartPick Hub server started")
        print("PASS - Homepage file:", (BASE / "index.html").exists())
        print("PASS - CSS file:", (BASE / "style.css").exists())
        print("PASS - JavaScript file:", (BASE / "script.js").exists())
        print("\nLOCAL PREVIEW: http://localhost:8000")

    except OSError as e:
        print("Server could not start:", e)
```

```python
from pathlib import Path
from bs4 import BeautifulSoup
from datetime import datetime

BASE = Path("/content/SmartPick-Hub")
FILE = BASE / "index.html"

html = FILE.read_text(encoding="utf-8")
soup = BeautifulSoup(html, "html.parser")

# Safety backup
backup = BASE / (
    "index-before-navigation-fix-"
    + datetime.now().strftime("%Y%m%d-%H%M%S")
    + ".html"
)
backup.write_text(html, encoding="utf-8")

nav = soup.find("nav", id="site-nav")

if not nav:
    print("FAIL - Homepage navigation not found")

else:
    mapping = {
        "Home": "index.html",
        "Technology": "technology.html",
        "Featured": "featured.html",
        "Buying Guides": "guides.html",
        "About": "about.html",
        "Explore Picks →": "featured.html"
    }

    for link in nav.find_all("a"):
        label = link.get_text(" ", strip=True)

        if label in mapping:
            link["href"] = mapping[label]

    FILE.write_text(
        str(soup),
        encoding="utf-8"
    )

    # Verify saved links
    check = BeautifulSoup(
        FILE.read_text(encoding="utf-8"),
        "html.parser"
    )

    nav = check.find("nav", id="site-nav")

    print("SMARTPICK HUB HOMEPAGE NAVIGATION TEST")
    print("-" * 45)

    for label, expected in mapping.items():
        link = next(
            (
                a for a in nav.find_all("a")
                if a.get_text(" ", strip=True) == label
            ),
            None
        )

        passed = link and link.get("href") == expected

        print(
            ("PASS" if passed else "FAIL"),
            "-",
            label,
            "=>",
            link.get("href") if link else "NOT FOUND"
        )

    print("\nNavigation fix saved successfully.")
```

```python
from pathlib import Path
from html.parser import HTMLParser
import re

file_path = Path("/content/SmartPick-Hub/featured.html")

if not file_path.exists():
    print("ERROR: featured.html file nahi mili.")
else:
    html = file_path.read_text(encoding="utf-8")

    class PageAudit(HTMLParser):
        def __init__(self):
            super().__init__()
            self.headings = []
            self.links = []
            self.images = 0
            self.in_heading = False
            self.heading_text = ""

        def handle_starttag(self, tag, attrs):
            attrs = dict(attrs)

            if tag in ("h1", "h2", "h3"):
                self.in_heading = True
                self.heading_text = ""

            if tag == "a" and attrs.get("href"):
                self.links.append(attrs["href"])

            if tag == "img":
                self.images += 1

        def handle_data(self, data):
            if self.in_heading:
                self.heading_text += data

        def handle_endtag(self, tag):
            if tag in ("h1", "h2", "h3") and self.in_heading:
                text = self.heading_text.strip()
                if text:
                    self.headings.append(text)
                self.in_heading = False

    audit = PageAudit()
    audit.feed(html)

    main = re.search(r"<main\b[^>]*>(.*?)</main>", html, re.I | re.S)
    main_content = main.group(1) if main else ""

    title = re.search(r"<title[^>]*>(.*?)</title>", html, re.I | re.S)

    print("SMARTPICK HUB FEATURED PAGE AUDIT")
    print("-" * 40)
    print("Page title:", title.group(1).strip() if title else "MISSING")
    print("Main section:", "PASS" if main else "MISSING")
    print("Content characters:", len(main_content))
    print("Headings:")
    for heading in audit.headings:
        print("-", heading)

    print("Images:", audit.images)
    print("Links:", len(audit.links))
    print("\nLinks found:")
    for link in audit.links:
        print("-", link)
```

```python
from pathlib import Path
from html.parser import HTMLParser
import re

file_path = Path("/content/SmartPick-Hub/guides.html")
html = file_path.read_text(encoding="utf-8")

class GuideAudit(HTMLParser):
    def __init__(self):
        super().__init__()
        self.headings = []
        self.links = []
        self.in_heading = False
        self.heading_text = ""

    def handle_starttag(self, tag, attrs):
        attrs = dict(attrs)
        if tag in ("h1", "h2", "h3"):
            self.in_heading = True
            self.heading_text = ""
        if tag == "a" and attrs.get("href"):
            self.links.append(attrs["href"])

    def handle_data(self, data):
        if self.in_heading:
            self.heading_text += data

    def handle_endtag(self, tag):
        if tag in ("h1", "h2", "h3") and self.in_heading:
            if self.heading_text.strip():
                self.headings.append(self.heading_text.strip())
            self.in_heading = False

audit = GuideAudit()
audit.feed(html)

main = re.search(r"<main\b[^>]*>(.*?)</main>", html, re.I | re.S)
title = re.search(r"<title[^>]*>(.*?)</title>", html, re.I | re.S)

print("SMARTPICK HUB BUYING GUIDES AUDIT")
print("-" * 40)
print("Page title:", title.group(1).strip() if title else "MISSING")
print("Main section:", "PASS" if main else "MISSING")
print("Content characters:", len(main.group(1)) if main else 0)
print("Headings:")
for heading in audit.headings:
    print("-", heading)

print("Links:", len(audit.links))
print("\nLinks found:")
for link in audit.links:
    print("-", link)
```

```python
from pathlib import Path
from html.parser import HTMLParser
from urllib.parse import urlparse, unquote

PROJECT = Path("/content/SmartPick-Hub")

required_pages = [
    "index.html",
    "technology.html",
    "featured.html",
    "guides.html",
    "about.html",
    "affiliate-disclosure.html"
]

class LinkParser(HTMLParser):
    def __init__(self):
        super().__init__()
        self.links = []

    def handle_starttag(self, tag, attrs):
        if tag == "a":
            href = dict(attrs).get("href")
            if href:
                self.links.append(href)

print("SMARTPICK HUB FINAL LINK AUDIT")
print("=" * 50)

print("\n1. REQUIRED PAGES")
for page in required_pages:
    status = "PASS" if (PROJECT / page).is_file() else "FAIL"
    print(f"{status} - {page}")

print("\n2. INTERNAL LINK CHECK")
total_links = 0
broken_links = []

for page in required_pages:
    file_path = PROJECT / page

    if not file_path.is_file():
        continue

    parser = LinkParser()
    parser.feed(file_path.read_text(encoding="utf-8"))

    for href in parser.links:
        parsed = urlparse(href)

        if parsed.scheme or parsed.netloc or href.startswith("#"):
            continue

        target = unquote(parsed.path)

        if not target:
            target = page

        total_links += 1
        target_path = (PROJECT / target).resolve()

        if not target_path.is_relative_to(PROJECT.resolve()):
            broken_links.append((page, href))
        elif not target_path.exists():
            broken_links.append((page, href))

if broken_links:
    for source, target in broken_links:
        print(f"FAIL - {source} -> {target}")
else:
    print("PASS - No broken internal links found")

print("\n3. AUDIT SUMMARY")
print("Pages checked:", len(required_pages))
print("Internal links checked:", total_links)
print("Broken links:", len(broken_links))

if all((PROJECT / p).is_file() for p in required_pages) and not broken_links:
    print("\nLINK AUDIT PASSED")
else:
    print("\nLINK AUDIT NEEDS FIXES")
```

```python
from pathlib import Path
import re

PROJECT = Path("/content/SmartPick-Hub")
CSS_FILE = PROJECT / "style.css"

pages = [
    "index.html",
    "technology.html",
    "featured.html",
    "guides.html",
    "about.html",
    "affiliate-disclosure.html"
]

print("SMARTPICK HUB CSS & MOBILE DESIGN AUDIT")
print("=" * 55)

if not CSS_FILE.exists():
    print("FAIL - style.css not found")
else:
    css = CSS_FILE.read_text(encoding="utf-8")
    print("PASS - style.css exists")
    print("CSS characters:", len(css))

    required_selectors = [
        "body",
        ".hero",
        ".hero-content",
        ".hero-visual",
        ".section",
        ".section-heading",
        ".card-grid",
        ".category-card",
        ".topic-card",
        ".btn",
        ".footer"
    ]

    print("\n1. REQUIRED CSS SELECTORS")
    for selector in required_selectors:
        pattern = re.escape(selector) + r"(?![\w-])"
        status = "PASS" if re.search(pattern, css) else "MISSING"
        print(f"{status} - {selector}")

    print("\n2. RESPONSIVE BREAKPOINTS")
    media_queries = re.findall(
        r"@media[^{]*",
        css,
        flags=re.I
    )

    if media_queries:
        for query in media_queries:
            print("PASS -", query.strip())
    else:
        print("MISSING - No responsive media queries found")

    print("\n3. HTML CSS CONNECTION")
    for page in pages:
        file_path = PROJECT / page

        if not file_path.exists():
            print(f"FAIL - {page} missing")
            continue

        html = file_path.read_text(encoding="utf-8")
        connected = re.search(
            r'<link\b[^>]*href=["\']style\.css["\']',
            html,
            re.I
        )

        print(
            f"{'PASS' if connected else 'FAIL'} - {page}"
        )

    print("\n4. CSS AUDIT COMPLETE")
```

```python
auth_js = """
const supabaseClient = window.supabase.createClient(
    window.SUPABASE_URL,
    window.SUPABASE_PUBLISHABLE_KEY
);

console.log("SmartPick Hub: Supabase connected!");

// SIGNUP
const signupForm = document.getElementById("signup-form");

if (signupForm) {
    signupForm.addEventListener("submit", async function(event) {
        event.preventDefault();

        const fullName = document.getElementById("full-name").value.trim();
        const email = document.getElementById("email").value.trim();
        const password = document.getElementById("password").value;
        const confirmPassword = document.getElementById("confirm-password").value;
        const message = document.getElementById("signup-message");

        if (password !== confirmPassword) {
            message.textContent = "Passwords do not match.";
            return;
        }

        message.textContent = "Creating your account...";

        const { error } = await supabaseClient.auth.signUp({
            email,
            password,
            options: {
                data: { full_name: fullName }
            }
        });

        message.textContent = error
            ? error.message
            : "Account created! Please check your email for verification.";
    });
}

// LOGIN
const loginForm = document.getElementById("login-form");

if (loginForm) {
    loginForm.addEventListener("submit", async function(event) {
        event.preventDefault();

        const email = document.getElementById("email").value.trim();
        const password = document.getElementById("password").value;
        const message = document.getElementById("login-message");

        message.textContent = "Logging in...";

        const { data, error } =
            await supabaseClient.auth.signInWithPassword({
                email,
                password
            });

        if (error) {
            message.textContent = error.message;
            return;
        }

        message.textContent = "Login successful! Welcome to SmartPick Hub.";
    });
}
"""

path = "/content/SmartPick-Hub/auth/auth.js"

with open(path, "w", encoding="utf-8") as f:
    f.write(auth_js)

print("PASS - Login and Signup functionality added")
```

```python
from pathlib import Path
import shutil

file_path = Path("/content/SmartPick-Hub/auth/signup.html")

# 1. Create backup
backup_path = file_path.with_name("signup-before-strong-password.html")
shutil.copy2(file_path, backup_path)

# 2. Read existing HTML
html = file_path.read_text(encoding="utf-8")

old_section = '''            <label for="password">Password</label>
            <input type="password" id="password" minlength="8" required>

            <label for="confirm-password">Confirm Password</label>
            <input type="password" id="confirm-password" minlength="8" required>'''

new_section = '''            <label for="password">Password</label>
            <input type="password" id="password" minlength="8" autocomplete="new-password" required>

            <button type="button" id="generate-password">Suggest Strong Password</button>
            <p id="password-strength" role="status">Password strength: Not entered</p>
            <p id="password-requirements">
                Use at least 12 characters, including uppercase, lowercase,
                a number and a special character.
            </p>

            <label for="confirm-password">Confirm Password</label>
            <input type="password" id="confirm-password" minlength="8" autocomplete="new-password" required>'''

if old_section not in html:
    print("STOP: Password section did not match. File was not changed.")
else:
    html = html.replace(old_section, new_section, 1)

    extra_script = '''
    <script>
    const passwordInput = document.getElementById("password");
    const strengthMessage = document.getElementById("password-strength");
    const generateButton = document.getElementById("generate-password");

    function checkPasswordStrength() {
        const password = passwordInput.value;
        let score = 0;

        if (password.length >= 12) score++;
        if (/[A-Z]/.test(password)) score++;
        if (/[a-z]/.test(password)) score++;
        if (/[0-9]/.test(password)) score++;
        if (/[^A-Za-z0-9]/.test(password)) score++;

        let strength = "Weak";
        if (score >= 4) strength = "Strong";
        else if (score >= 3) strength = "Medium";

        strengthMessage.textContent = password
            ? "Password strength: " + strength
            : "Password strength: Not entered";
    }

    passwordInput.addEventListener("input", checkPasswordStrength);

    generateButton.addEventListener("click", function() {
        const chars = "ABCDEFGHJKLMNPQRSTUVWXYZabcdefghijkmnopqrstuvwxyz23456789!@#$%&*?";
        const randomValues = new Uint32Array(16);
        crypto.getRandomValues(randomValues);

        let password = "";
        for (const value of randomValues) {
            password += chars[value % chars.length];
        }

        passwordInput.value = password;
        passwordInput.dispatchEvent(new Event("input"));
        document.getElementById("confirm-password").value = password;
    });
    </script>
    '''

    html = html.replace(
        '    <script src="auth.js"></script>',
        extra_script + '\\n    <script src="auth.js"></script>',
        1
    )

    file_path.write_text(html, encoding="utf-8")
    print("SUCCESS: Strong password feature added!")
    print("Backup saved:", backup_path)
    print("Updated file:", file_path)
```

```python
from pathlib import Path

file_path = Path("/content/SmartPick-Hub/auth/signup.html")
html = file_path.read_text(encoding="utf-8")

old = '''<input type="password" id="password" minlength="8" autocomplete="new-password" required>'''

new = '''<div class="password-field">
                <input type="password" id="password" minlength="8" autocomplete="new-password" required>
                <button type="button" id="toggle-password">Show</button>
            </div>'''

old_confirm = '''<input type="password" id="confirm-password" minlength="8" autocomplete="new-password" required>'''

new_confirm = '''<div class="password-field">
                <input type="password" id="confirm-password" minlength="8" autocomplete="new-password" required>
                <button type="button" id="toggle-confirm-password">Show</button>
            </div>'''

if old not in html or old_confirm not in html:
    print("STOP: Expected password fields not found. No changes made.")
else:
    html = html.replace(old, new, 1)
    html = html.replace(old_confirm, new_confirm, 1)

    toggle_script = '''
    <script>
    function setupPasswordToggle(inputId, buttonId) {
        const input = document.getElementById(inputId);
        const button = document.getElementById(buttonId);

        button.addEventListener("click", function() {
            const isHidden = input.type === "password";
            input.type = isHidden ? "text" : "password";
            button.textContent = isHidden ? "Hide" : "Show";
        });
    }

    setupPasswordToggle("password", "toggle-password");
    setupPasswordToggle("confirm-password", "toggle-confirm-password");
    </script>
    '''

    html = html.replace(
        '    <script src="auth.js"></script>',
        toggle_script + '\\n    <script src="auth.js"></script>',
        1
    )

    file_path.write_text(html, encoding="utf-8")
    print("SUCCESS: Show/Hide password feature added!")
```

```python
from pathlib import Path
import shutil

auth_file = Path("/content/SmartPick-Hub/auth/auth.js")

# Backup existing code
backup = auth_file.with_name("auth-before-dashboard-profile.js")
shutil.copy2(auth_file, backup)

js = auth_file.read_text(encoding="utf-8")

# Add dashboard functionality
dashboard_code = """

// DASHBOARD
const dashboardContent = document.getElementById("dashboard-content");
const dashboardMessage = document.getElementById("dashboard-message");

if (dashboardMessage) {
    (async function() {
        const { data: { user }, error } =
            await supabaseClient.auth.getUser();

        if (error || !user) {
            window.location.replace("login.html");
            return;
        }

        const name = user.user_metadata?.full_name || "SmartPick Hub User";

        document.getElementById("user-name").textContent = name;
        document.getElementById("user-email").textContent = user.email;

        const { data: profile } = await supabaseClient
            .from("profiles")
            .select("role")
            .eq("id", user.id)
            .maybeSingle();

        document.getElementById("user-role").textContent =
            profile?.role || "customer";

        dashboardContent.style.display = "block";
        dashboardMessage.textContent = "You are securely logged in.";
    })();
}

// LOGOUT
const logoutButton = document.getElementById("logout-button");

if (logoutButton) {
    logoutButton.addEventListener("click", async function() {
        logoutButton.disabled = true;
        logoutButton.textContent = "Logging out...";

        const { error } = await supabaseClient.auth.signOut();

        if (error) {
            alert("Logout failed. Please try again.");
            logoutButton.disabled = false;
            logoutButton.textContent = "Logout";
            return;
        }

        window.location.replace("login.html");
    });
}
"""

if "// DASHBOARD" not in js:
    js += dashboard_code
    auth_file.write_text(js, encoding="utf-8")
    print("SUCCESS: Dashboard profile and logout logic added!")
else:
    print("Dashboard logic already exists. No changes made.")

print("Backup:", backup)
```

```python
import json
from pathlib import Path

notebook = Path(
    "/content/drive/MyDrive/Colab Notebooks/SmartPick_Hub_Development.ipynb"
)

data = json.loads(notebook.read_text(encoding="utf-8"))

write_keywords = [
    "write_text",
    "write_bytes",
    "open(",
    ".write(",
    "shutil.copy",
    "copyfile",
    "copytree",
    "make_archive",
    "ZipFile",
    "os.makedirs",
    "mkdir("
]

print("POSSIBLE FILE-WRITING CELLS")
print("=" * 50)

for i, cell in enumerate(data["cells"][:95], start=1):
    if cell.get("cell_type") != "code":
        continue

    lines = "".join(cell.get("source", [])).splitlines()
    matches = [
        (n, line.strip())
        for n, line in enumerate(lines, start=1)
        if any(keyword in line for keyword in write_keywords)
    ]

    if matches:
        print(f"\nCELL {i}")
        for line_no, line in matches:
            print(f"  Line {line_no}: {line[:200]}")

print("\nSCAN COMPLETED")
```

```python
import json
import re
from pathlib import Path
from google.colab import files

source = Path(
    "/content/drive/MyDrive/Colab Notebooks/"
    "SmartPick_Hub_Development_RECOVERY_COPY.ipynb"
)

output = source.with_name(
    "SmartPick_Hub_Development_REVIEW_REDACTED.ipynb"
)

data = json.loads(source.read_text(encoding="utf-8"))

# Notebook source aur outputs se Publishable Key redact karein
for cell in data.get("cells", []):
    if "source" in cell:
        cell["source"] = [
            re.sub(
                r"sb_publishable_[A-Za-z0-9_-]+",
                "REDACTED_PUBLISHABLE_KEY",
                line
            )
            for line in cell["source"]
        ]

    for item in cell.get("outputs", []):
        if "text" in item:
            item["text"] = [
                re.sub(
                    r"sb_publishable_[A-Za-z0-9_-]+",
                    "REDACTED_PUBLISHABLE_KEY",
                    line
                )
                for line in item["text"]
            ]

output.write_text(
    json.dumps(data, indent=1, ensure_ascii=False),
    encoding="utf-8"
)

print("Review copy created:", output.name)
files.download(str(output))
```

```python
from pathlib import Path
from bs4 import BeautifulSoup
from copy import deepcopy
import shutil

# Existing recovered website
SOURCE = Path("/content/SmartPick-Hub-RECOVERY")

# Separate multipage staging folder
PROJECT = Path("/content/SmartPick-Hub-MULTIPAGE")
PROJECT.mkdir(parents=True, exist_ok=True)

# Read recovered homepage
home_file = SOURCE / "index.html"

original = BeautifulSoup(
    home_file.read_text(encoding="utf-8"),
    "html.parser"
)

# Section mapping
page_sections = {
    "index.html": (".hero", "SmartPick Hub | Smart Choices. Better Living."),
    "technology.html": ("#technology", "Technology & AI Tools | SmartPick Hub"),
    "featured.html": ("#featured", "Featured Topics | SmartPick Hub"),
    "guides.html": ("#guides", "Buying Guides | SmartPick Hub"),
    "about.html": ("#about", "About Us | SmartPick Hub"),
    "affiliate-disclosure.html": (
        ".disclosure",
        "Affiliate Disclosure | SmartPick Hub"
    )
}

# Navigation destinations
nav_links = {
    "Home": "index.html",
    "Technology": "technology.html",
    "Featured": "featured.html",
    "Buying Guides": "guides.html",
    "About": "about.html",
    "Explore Picks →": "featured.html"
}

# Generate pages
for filename, (selector, title) in page_sections.items():

    page = BeautifulSoup(str(original), "html.parser")

    # Set page title
    if page.title:
        page.title.string = title

    # Update navigation links
    nav = page.find("nav")

    if nav:
        for link in nav.find_all("a"):
            label = link.get_text(" ", strip=True)

            if label in nav_links:
                link["href"] = nav_links[label]

                if nav_links[label] == filename:
                    link["aria-current"] = "page"
                else:
                    link.attrs.pop("aria-current", None)

    # Update footer links
    footer = page.find("footer")

    if footer:
        for link in footer.find_all("a"):
            label = link.get_text(" ", strip=True)

            if label in nav_links:
                link["href"] = nav_links[label]

    # Find required section
    section = original.select_one(selector)

    if section is None:
        raise ValueError(
            f"Section missing for {filename}: {selector}"
        )

    # Create new main containing only selected section
    new_main = page.new_tag("main")
    selected_section = deepcopy(section)

    # Give Hero and Disclosure sections permanent IDs
    if filename == "index.html":
        selected_section["id"] = "home"

    if filename == "affiliate-disclosure.html":
        selected_section["id"] = "affiliate-disclosure"

    new_main.append(selected_section)

    old_main = page.find("main")

    if old_main is None:
        raise ValueError("Main element missing in source HTML")

    old_main.replace_with(new_main)

    # Save generated page
    output = PROJECT / filename

    output.write_text(
        "<!DOCTYPE html>\n" + str(page),
        encoding="utf-8"
    )

    print("PASS - Created:", filename)

# Copy CSS and JavaScript
for filename in ["style.css", "script.js"]:
    shutil.copy2(SOURCE / filename, PROJECT / filename)
    print("PASS - Copied:", filename)

print("\nMULTIPAGE GENERATION COMPLETE")
print("Staging folder:", PROJECT)
print("Total HTML pages:", len(page_sections))
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub-MULTIPAGE")

expected_pages = {
    "index.html": "home",
    "technology.html": "technology",
    "featured.html": "featured",
    "guides.html": "guides",
    "about.html": "about",
    "affiliate-disclosure.html": "affiliate-disclosure"
}

print("SMARTPICK HUB MULTI-PAGE QUALITY AUDIT")
print("=" * 55)

total_pass = 0
total_checks = 0

for filename, expected_id in expected_pages.items():

    path = PROJECT / filename
    soup = BeautifulSoup(
        path.read_text(encoding="utf-8"),
        "html.parser"
    )

    checks = {
        "Title": soup.title is not None,
        "Header": soup.find("header") is not None,
        "Navigation": soup.find("nav") is not None,
        "Main content": (
            soup.main is not None
            and bool(soup.main.get_text(" ", strip=True))
        ),
        "Correct section": (
            soup.main is not None
            and soup.main.find(id=expected_id) is not None
        ),
        "Footer": soup.find("footer") is not None,
        "CSS connected": soup.find("link", href="style.css") is not None,
        "JavaScript connected": soup.find("script", src="script.js") is not None
    }

    passed = sum(checks.values())
    total_pass += passed
    total_checks += len(checks)

    print(f"\n{filename}: {passed}/{len(checks)}")

    for name, result in checks.items():
        print(("PASS" if result else "FAIL"), "-", name)

    # Navigation destination check
    nav = soup.find("nav")
    if nav:
        for link in nav.find_all("a", href=True):
            href = link["href"]

            if href.endswith(".html"):
                target = PROJECT / href

                if not target.exists():
                    print("BROKEN NAV LINK:", href)

print("\n" + "=" * 55)
print("TOTAL:", total_pass, "/", total_checks)
print("AUDIT FINISHED")
```

```python
from pathlib import Path
from IPython.display import HTML, display
import re

PROJECT = Path("/content/SmartPick-Hub-MULTIPAGE")

html = (PROJECT / "index.html").read_text(encoding="utf-8")
css = (PROJECT / "style.css").read_text(encoding="utf-8")
js = (PROJECT / "script.js").read_text(encoding="utf-8")

# CSS ko HTML ke andar load karna
html = re.sub(
    r'<link\b[^>]*href=["\']style\.css["\'][^>]*>',
    lambda m: "<style>\n" + css + "\n</style>",
    html,
    flags=re.IGNORECASE
)

# JavaScript ko HTML ke andar load karna
html = re.sub(
    r'<script\b[^>]*src=["\']script\.js["\'][^>]*>\s*</script>',
    lambda m: "<script>\n" + js + "\n</script>",
    html,
    flags=re.IGNORECASE
)

display(HTML(html))
```

```python
from pathlib import Path
from bs4 import BeautifulSoup

PROJECT = Path("/content/SmartPick-Hub-PROFESSIONAL-RESTORED")
FILE = PROJECT / "index.html"

# Read existing homepage
html = FILE.read_text(encoding="utf-8")
soup = BeautifulSoup(html, "html.parser")

# Update navigation links only
navigation = soup.find("nav")

if navigation:
    replacements = {
        "Home": "index.html",
        "Technology": "technology.html",
        "Featured": "featured.html",
        "Buying Guides": "guides.html",
        "About": "about.html"
    }

    for link in navigation.find_all("a"):
        label = link.get_text(" ", strip=True).replace("→", "").strip()

        if label in replacements:
            link["href"] = replacements[label]

# Update homepage footer links
footer = soup.find("footer")

if footer:
    replacements = {
        "Home": "index.html",
        "Technology": "technology.html",
        "Featured": "featured.html",
        "Buying Guides": "guides.html",
        "About": "about.html",
        "Affiliate Disclosure": "affiliate-disclosure.html"
    }

    for link in footer.find_all("a"):
        label = link.get_text(" ", strip=True).replace("→", "").strip()

        if label in replacements:
            link["href"] = replacements[label]

# Save updated homepage
FILE.write_text(str(soup), encoding="utf-8")

# Verify navigation
print("HOMEPAGE NAVIGATION CHECK")
for link in soup.find("nav").find_all("a"):
    print(link.get_text(" ", strip=True), "->", link.get("href"))

print("\nHomepage navigation updated successfully!")
```

```python

from pathlib import Path
from datetime import datetime
import shutil

PROJECT = Path("/content/SmartPick-Hub-PROFESSIONAL-RESTORED")
DRIVE_BACKUP = Path("/content/drive/MyDrive")

technology = PROJECT / "technology.html"
featured = PROJECT / "featured.html"

# Check existing pages
for file in [technology, featured]:
    print(file.name, "Exists:", file.exists())

# Create a safety backup before making changes
backup_name = (
    "SmartPick-Hub-Before-Guides-"
    + datetime.now().strftime("%Y%m%d_%H%M%S")
    + ".zip"
)

backup_path = shutil.make_archive(
    str(DRIVE_BACKUP / backup_name.replace(".zip", "")),
    "zip",
    root_dir=PROJECT
)

print("\nBackup created:")
print(backup_path)

# Inspect existing page structure
from bs4 import BeautifulSoup

for file in [technology, featured]:
    html = file.read_text(encoding="utf-8")
    soup = BeautifulSoup(html, "html.parser")

    print(f"\n--- {file.name} STRUCTURE ---")

    for tag in ["header", "nav", "main", "section", "footer"]:
        elements = soup.find_all(tag)
        print(f"{tag}: {len(elements)}")

    main = soup.find("main")
    if main:
        print("Main classes:", main.get("class"))
        print("Main direct children:")
        for child in main.find_all(recursive=False):
            print(" ", child.name, child.get("class"))

    print("CSS:", [
        link.get("href")
        for link in soup.find_all("link", rel="stylesheet")
    ])
```

```python
from pathlib import Path
from datetime import datetime
import shutil
import zipfile

DRIVE = Path("/content/drive/MyDrive")
NOTEBOOK = DRIVE / "Colab Notebooks/SmartPick_Hub_Development.ipynb"
PROJECT = Path("/content/SmartPick-Hub-PROFESSIONAL-RESTORED")

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
BACKUP = DRIVE / f"SmartPick-Hub-Final-Backup-{timestamp}"
BACKUP.mkdir(parents=True)

# 1. Save a separate notebook copy
notebook_copy = BACKUP / "SmartPick_Hub_Development.ipynb"
shutil.copy2(NOTEBOOK, notebook_copy)

# 2. Save complete website project
project_copy = BACKUP / "SmartPick-Hub-PROFESSIONAL-RESTORED"
shutil.copytree(PROJECT, project_copy)

# 3. Create ZIP
zip_path = DRIVE / f"SmartPick-Hub-Final-Backup-{timestamp}.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for file in BACKUP.rglob("*"):
        if file.is_file():
            z.write(file, file.relative_to(BACKUP))

# 4. Verify
with zipfile.ZipFile(zip_path, "r") as z:
    integrity = z.testzip()

print("NOTEBOOK BACKUP:", notebook_copy.exists())
print("PROJECT BACKUP:", project_copy.exists())
print("PROJECT FILES:", sum(1 for p in project_copy.rglob("*") if p.is_file()))
print("ZIP BACKUP:", zip_path.exists())
print("ZIP INTEGRITY:", "PASS" if integrity is None else "FAILED")
print("\nBACKUP LOCATION:", BACKUP)
print("ZIP LOCATION:", zip_path)

assert integrity is None
print("\nSUCCESS: NOTEBOOK AND PROJECT BACKED UP!")
```

```python
from pathlib import Path
import zipfile

print("NOTEBOOK FILES IN /content:\n")

for path in Path("/content").glob("*.ipynb"):
    print(path.name, "|", path.stat().st_size, "bytes")

print("\nNOTEBOOKS INSIDE ZIP BACKUPS:\n")

for zip_path in Path("/content").glob("*.zip"):
    try:
        with zipfile.ZipFile(zip_path, "r") as z:
            notebooks = [
                name for name in z.namelist()
                if name.lower().endswith(".ipynb")
            ]
            if notebooks:
                print(zip_path.name)
                for name in notebooks:
                    print("  ", name)
    except zipfile.BadZipFile:
        print("Invalid ZIP:", zip_path.name)
```

```python
import json
from pathlib import Path

BASE = Path("/content/gdrive/MyDrive")

files = [
    BASE / "Colab Notebooks/SmartPick_Hub_Development.ipynb",
    BASE / "Colab Notebooks/SmartPick_Hub_Development_RECOVERY_COPY.ipynb",
    BASE / "Colab Notebooks/SmartPick_Hub_Development_REVIEW_REDACTED.ipynb",
    BASE / "Colab Notebooks/Copy of SmartPick_Hub_Development.ipynb",
    BASE / "SmartPick-Hub-Final-Backup-20261109_141914/SmartPick_Hub_Development.ipynb"
]

for path in files:
    print("\n" + "=" * 55)
    print("FILE:", path.name)

    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    cells = data.get("cells", [])
    code_cells = [c for c in cells if c.get("cell_type") == "code"]

    print("Total cells:", len(cells))
    print("Code cells:", len(code_cells))
    print("Markdown cells:", len(cells) - len(code_cells))
    print("Last 3 cells:")

    for cell in cells[-3:]:
        source = "".join(cell.get("source", []))
        print("-", cell.get("cell_type"), ":", source[:180].replace("\n", " "))
```

```python
from pathlib import Path
import shutil
import zipfile

SOURCE = Path(
    "/content/gdrive/MyDrive/Colab Notebooks/"
    "SmartPick_Hub_Development.ipynb"
)

BACKUP_DIR = Path(
    "/content/gdrive/MyDrive/SmartPick-Hub-Notebook-Backup-RECOVERED"
)

BACKUP_DIR.mkdir(exist_ok=True)

COPY = BACKUP_DIR / "SmartPick_Hub_Development_231_CELLS.ipynb"
ZIP = BACKUP_DIR / "SmartPick_Hub_Development_231_CELLS.zip"

if not SOURCE.exists():
    print("ERROR: Source notebook not found.")
else:
    if not COPY.exists():
        shutil.copy2(SOURCE, COPY)
        print("Notebook copy created.")
    else:
        print("Notebook copy already exists. Not overwritten.")

    if not ZIP.exists():
        with zipfile.ZipFile(ZIP, "w", zipfile.ZIP_DEFLATED) as z:
            z.write(COPY, arcname=COPY.name)
        print("ZIP backup created.")
    else:
        print("ZIP already exists. Not overwritten.")

    print("\nFINAL CHECK")
    print("Notebook copy:", COPY.exists())
    print("ZIP copy:", ZIP.exists())

    with zipfile.ZipFile(ZIP, "r") as z:
        print("ZIP integrity:", z.testzip() or "PASS")
```

```python
from pathlib import Path

PROJECT = Path("/content/SmartPick-Hub")
css = (PROJECT / "style.css").read_text(encoding="utf-8")

keywords = [
    "nav{",
    "nav {",
    "#site-nav",
    ".nav-cta",
    ".menu-toggle",
    ".menu-open",
    "@media"
]

for keyword in keywords:
    print("\n" + "=" * 45)
    print("SEARCH:", keyword)

    lines = css.splitlines()
    matches = [
        (i, line.strip())
        for i, line in enumerate(lines)
        if keyword in line
    ]

    for i, line in matches[:15]:
        print(f"Line {i+1}: {line[:300]}")
```

```python
from pathlib import Path

PROJECT = Path("/content/SmartPick-Hub")
file = PROJECT / "index.html"

html = file.read_text(encoding="utf-8")

old_nav = '''<nav id="site-nav">
<a href="index.html">Home</a>
<a href="technology.html">Technology</a>
<a href="featured.html">Featured</a>
<a href="guides.html">Buying Guides</a>
<a href="about.html">About</a>
<a class="nav-cta" href="featured.html">Explore Picks →</a>
</nav>'''

new_nav = '''<nav id="site-nav">
<a href="index.html">Home</a>
<a href="technology.html">Technology</a>
<a href="featured.html">Featured</a>
<a href="guides.html">Buying Guides</a>
<a href="about.html">About</a>

<a href="auth.html" id="login-nav-link">Login / Sign Up</a>
<a href="subscribe.html" id="subscribe-nav-link">Subscribe</a>
<a href="profile.html" id="profile-nav-link" hidden>Profile</a>

<a class="nav-cta" href="featured.html">Explore Picks →</a>
</nav>'''

if old_nav not in html:
    print("ERROR: Original navigation not found. No changes made.")
else:
    updated_html = html.replace(old_nav, new_nav, 1)
    file.write_text(updated_html, encoding="utf-8")
    print("Homepage navigation updated successfully!")

    check = file.read_text(encoding="utf-8")
    print("Login link:", 'id="login-nav-link"' in check)
    print("Subscribe link:", 'id="subscribe-nav-link"' in check)
    print("Profile link:", 'id="profile-nav-link"' in check)
    print("Profile initially hidden:", 'id="profile-nav-link" hidden' in check)
```

```python
from pathlib import Path
import requests

PROJECT = Path("/content/SmartPick-Hub")
content = (PROJECT / "supabase-config.js").read_text(encoding="utf-8")

url = next(
    line.split("=", 1)[1].strip().strip('"; ')
    for line in content.splitlines()
    if "const SUPABASE_URL" in line
)

key = next(
    line.split("=", 1)[1].strip().rstrip(";").strip().strip('"').strip("'")
    for line in content.splitlines()
    if "const SUPABASE_PUBLISHABLE_KEY" in line
)

try:
    response = requests.get(
        url + "/auth/v1/settings",
        headers={"apikey": key},
        timeout=20
    )

    print("HTTP status:", response.status_code)

    if response.status_code == 200:
        print("SUCCESS: Supabase connection working!")
    elif response.status_code in [401, 403]:
        print("Key rejected. Please check the key in your config file.")
    else:
        print("Connection response received. Status:", response.status_code)

except Exception as error:
    print("Connection test error:", type(error).__name__)
```

```python
from pathlib import Path

PROJECT = Path("/content/SmartPick-Hub")

auth_js = """
const supabaseClient = window.supabase.createClient(
    SUPABASE_URL,
    SUPABASE_PUBLISHABLE_KEY
);

const authForm = document.getElementById("auth-form");
const authTitle = document.getElementById("auth-title");
const authDescription = document.getElementById("auth-description");
const authSubmit = document.getElementById("auth-submit");
const authSwitch = document.getElementById("auth-switch");
const authSwitchText = document.getElementById("auth-switch-text");
const authMessage = document.getElementById("auth-message");

let isSignup = false;

authSwitch.addEventListener("click", () => {
    isSignup = !isSignup;

    authTitle.textContent = isSignup ? "Create Account" : "Welcome Back";
    authDescription.textContent = isSignup
        ? "Create your SmartPick Hub account."
        : "Login to your SmartPick Hub account.";

    authSubmit.textContent = isSignup ? "Sign Up" : "Login";
    authSwitchText.textContent = isSignup
        ? "Already have an account?"
        : "Don't have an account?";

    authSwitch.textContent = isSignup ? "Login" : "Sign Up";
    authMessage.textContent = "";
});

authForm.addEventListener("submit", async (event) => {
    event.preventDefault();

    const email = document.getElementById("auth-email").value.trim();
    const password = document.getElementById("auth-password").value;

    authSubmit.disabled = true;
    authMessage.textContent = "Please wait...";

    try {
        if (isSignup) {
            const { data, error } = await supabaseClient.auth.signUp({
                email,
                password,
                options: {
                    emailRedirectTo: window.location.origin + "/profile.html"
                }
            });

            if (error) throw error;

            authMessage.textContent = data.session
                ? "Account created successfully!"
                : "Account created! Please check your email to verify your account.";

        } else {
            const { data, error } = await supabaseClient.auth.signInWithPassword({
                email,
                password
            });

            if (error) throw error;

            authMessage.textContent = "Login successful! Opening your profile...";

           window.location.href = "index.html";
        }

    } catch (error) {
        authMessage.textContent = error.message;
    } finally {
        authSubmit.disabled = false;
    }
});
"""

file = PROJECT / "auth.js"
file.write_text(auth_js.strip(), encoding="utf-8")

print("auth.js created:", file.exists())
print("File size:", file.stat().st_size, "bytes")
```

```python
from pathlib import Path

PROJECT = Path("/content/SmartPick-Hub")

# Updated authentication HTML
auth_html = '''<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <meta name="description" content="Sign in or create your SmartPick Hub account.">
    <title>Login / Sign Up | SmartPick Hub</title>
    <link rel="stylesheet" href="style.css">
</head>
<body>

<header class="header">
    <div class="navbar">
        <a class="logo" href="index.html">SmartPick Hub</a>
        <a href="index.html">Back to Home</a>
    </div>
</header>

<main class="auth-container">
    <section class="auth-card">
        <h1 id="auth-title">Welcome Back</h1>
        <p id="auth-description">Login to your SmartPick Hub account.</p>

        <form id="auth-form">
            <label for="auth-email">Email Address</label>
            <input type="email" id="auth-email"
                   placeholder="Enter your email"
                   autocomplete="email" required>

            <label for="auth-password">Password</label>
            <div class="password-field">
                <input type="password" id="auth-password"
                       placeholder="Enter your password"
                       autocomplete="current-password"
                       minlength="8" required>
                <button type="button" class="password-toggle"
                        data-target="auth-password"
                        aria-label="Show password">Show</button>
            </div>

            <div id="confirm-password-group" hidden>
                <label for="auth-confirm-password">Confirm Password</label>
                <div class="password-field">
                    <input type="password" id="auth-confirm-password"
                           placeholder="Re-enter your password"
                           autocomplete="new-password"
                           minlength="8">
                    <button type="button" class="password-toggle"
                            data-target="auth-confirm-password"
                            aria-label="Show confirm password">Show</button>
                </div>
            </div>

            <button type="submit" id="auth-submit">Login</button>
        </form>

        <p id="auth-message" role="status" aria-live="polite"></p>

        <p>
            <span id="auth-switch-text">Don't have an account?</span>
            <button type="button" id="auth-switch">Sign Up</button>
        </p>

        <a href="index.html">Return to Homepage</a>
    </section>
</main>

<script src="https://cdn.jsdelivr.net/npm/@supabase/supabase-js@2"></script>
<script src="supabase-config.js"></script>
<script src="auth.js"></script>

</body>
</html>
'''

(PROJECT / "auth.html").write_text(auth_html, encoding="utf-8")

# Updated authentication JavaScript
auth_js = """
const supabaseClient = window.supabase.createClient(
    SUPABASE_URL,
    SUPABASE_PUBLISHABLE_KEY
);

const authForm = document.getElementById("auth-form");
const authTitle = document.getElementById("auth-title");
const authDescription = document.getElementById("auth-description");
const authSubmit = document.getElementById("auth-submit");
const authSwitch = document.getElementById("auth-switch");
const authSwitchText = document.getElementById("auth-switch-text");
const authMessage = document.getElementById("auth-message");

const passwordInput = document.getElementById("auth-password");
const confirmGroup = document.getElementById("confirm-password-group");
const confirmInput = document.getElementById("auth-confirm-password");

let isSignup = false;

function updateAuthMode() {
    authTitle.textContent = isSignup ? "Create Account" : "Welcome Back";

    authDescription.textContent = isSignup
        ? "Create your SmartPick Hub account."
        : "Login to your SmartPick Hub account.";

    authSubmit.textContent = isSignup ? "Sign Up" : "Login";

    authSwitchText.textContent = isSignup
        ? "Already have an account?"
        : "Don't have an account?";

    authSwitch.textContent = isSignup ? "Login" : "Sign Up";

    confirmGroup.hidden = !isSignup;
    confirmInput.required = isSignup;

    passwordInput.autocomplete = isSignup
        ? "new-password"
        : "current-password";

    authMessage.textContent = "";
}

authSwitch.addEventListener("click", () => {
    isSignup = !isSignup;
    authForm.reset();
    updateAuthMode();
});

document.querySelectorAll(".password-toggle").forEach(button => {
    button.addEventListener("click", () => {
        const input = document.getElementById(button.dataset.target);
        const showing = input.type === "text";

        input.type = showing ? "password" : "text";
        button.textContent = showing ? "Show" : "Hide";
        button.setAttribute(
            "aria-label",
            showing ? "Show password" : "Hide password"
        );
    });
});

authForm.addEventListener("submit", async (event) => {
    event.preventDefault();

    const email = document.getElementById("auth-email").value.trim();
    const password = passwordInput.value;

    if (isSignup && password !== confirmInput.value) {
        authMessage.textContent = "Passwords do not match. Please try again.";
        return;
    }

    authSubmit.disabled = true;
    authMessage.textContent = "Please wait...";

    try {
        if (isSignup) {
            const { data, error } = await supabaseClient.auth.signUp({
                email,
                password,
                options: {
                    emailRedirectTo: window.location.origin + "/profile.html"
                }
            });

            if (error) throw error;

            authMessage.textContent = data.session
                ? "Account created successfully!"
                : "Account created! Please check your email to verify your account.";

        } else {
            const { data, error } = await supabaseClient.auth.signInWithPassword({
                email,
                password
            });

            if (error) throw error;

            authMessage.textContent = "Login successful! Opening your profile...";
            window.location.href = "profile.html";
        }

    } catch (error) {
        authMessage.textContent = error.message;
    } finally {
        authSubmit.disabled = false;
    }
});

updateAuthMode();
"""

(PROJECT / "auth.js").write_text(auth_js.strip(), encoding="utf-8")

print("auth.html updated:", (PROJECT / "auth.html").exists())
print("auth.js updated:", (PROJECT / "auth.js").exists())
print("Confirm password added:", "auth-confirm-password" in auth_html)
print("Password show/hide added:", "password-toggle" in auth_html)
```

```python
import subprocess
import requests
import time
from pathlib import Path
from google.colab import output

# 1. Original project folder
PROJECT = Path("/content/SmartPick-Hub")
AUTH_FILE = PROJECT / "auth.html"

if not AUTH_FILE.exists():
    raise FileNotFoundError("STOP: auth.html project folder mein nahi mili.")

# 2. Naya port use karein
PORT = 8002

# 3. Naya preview server start karein
server = subprocess.Popen(
    [
        "python", "-m", "http.server",
        str(PORT),
        "--bind", "0.0.0.0",
        "--directory", str(PROJECT)
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

# 4. Server ready hone ka wait aur local test
for attempt in range(15):
    try:
        response = requests.get(
            f"http://127.0.0.1:{PORT}/auth.html",
            timeout=3
        )
        if response.status_code == 200:
            break
    except requests.RequestException:
        time.sleep(1)
else:
    raise RuntimeError("Preview server start nahi hua.")

# 5. Colab ka correct external preview URL
preview_url = output.eval_js(
    f"google.colab.kernel.proxyPort({PORT})"
).rstrip("/")

auth_url = preview_url + "/auth.html"

# 6. Final verification
print("LOCAL TEST:", response.status_code)
print("AUTH FILE:", AUTH_FILE.exists())
print("\nYOUR NEW AUTH PAGE LINK:")
print(auth_url)
```

```python
from pathlib import Path
from datetime import datetime
import shutil

PROJECT = Path("/content/SmartPick-Hub")

PAGES = [
    "index.html",
    "technology.html",
    "featured.html",
    "guides.html",
    "about.html",
    "affiliate-disclosure.html"
]

AUTH_NAV = PROJECT / "auth-nav.js"

# Safety checks: pehle tamam files verify hongi
for name in PAGES:
    if not (PROJECT / name).exists():
        raise FileNotFoundError(f"STOP: {name} missing hai.")

if not (PROJECT / "supabase-config.js").exists():
    raise FileNotFoundError("STOP: Supabase config missing hai.")

if AUTH_NAV.exists():
    raise FileExistsError(
        "STOP: auth-nav.js pehle se mojood hai. "
        "Existing file overwrite nahi ki."
    )

# Sab HTML changes pehle memory mein prepare karein
updated_pages = {}

nav_links = """
<a href="auth.html" id="login-nav-link">Login / Sign Up</a>
<a href="subscribe.html" id="subscribe-nav-link">Subscribe</a>
<a href="profile.html" id="profile-nav-link" hidden>Profile</a>
"""

script_tags = """<script src="https://cdn.jsdelivr.net/npm/@supabase/supabase-js@2"></script>
<script src="supabase-config.js"></script>
<script src="auth-nav.js"></script>
"""

for name in PAGES:
    path = PROJECT / name
    content = path.read_text(encoding="utf-8")

    if 'id="site-nav"' not in content:
        raise ValueError(f"STOP: {name} mein site-nav nahi mila.")

    # Login/Profile links missing hon to add karein
    if 'id="login-nav-link"' not in content:
        lines = content.splitlines(keepends=True)
        inserted = False

        for i, line in enumerate(lines):
            if 'class="nav-cta"' in line:
                lines.insert(i, nav_links)
                inserted = True
                break

        if not inserted:
            raise ValueError(f"STOP: {name} mein CTA location nahi mili.")

        content = "".join(lines)

    # Existing Explore Picks link ko preserve karein
    if 'class="nav-cta"' not in content:
        raise ValueError(f"STOP: {name} ka Explore Picks link missing hai.")

    # Shared authentication scripts add karein
    if "auth-nav.js" not in content:
        old_script = '<script src="script.js"></script>'

        if content.count(old_script) != 1:
            raise ValueError(
                f"STOP: {name} mein script.js ka expected location nahi mila."
            )

        content = content.replace(
            old_script,
            script_tags + old_script,
            1
        )

    updated_pages[name] = content

# Shared authentication navigation JavaScript
auth_nav_code = r"""
// SmartPick Hub - Shared Authentication Navigation

const smartPickNavClient = window.supabase.createClient(
    SUPABASE_URL,
    SUPABASE_PUBLISHABLE_KEY
);

function updateSmartPickNavigation(isLoggedIn) {
    const loginLink = document.getElementById("login-nav-link");
    const profileLink = document.getElementById("profile-nav-link");

    if (loginLink) {
        loginLink.hidden = isLoggedIn;
    }

    if (profileLink) {
        profileLink.hidden = !isLoggedIn;
    }
}

async function initializeSmartPickNavigation() {
    try {
        const { data, error } =
            await smartPickNavClient.auth.getSession();

        if (error) throw error;

        updateSmartPickNavigation(Boolean(data.session));

    } catch (error) {
        console.error("Navigation session check failed:", error.message);
        updateSmartPickNavigation(false);
    }
}

smartPickNavClient.auth.onAuthStateChange(
    (event, session) => {
        updateSmartPickNavigation(Boolean(session));
    }
);

initializeSmartPickNavigation();
"""

# Backup: original six HTML pages
timestamp = datetime.now().strftime("%Y%m%d-%H%M%S")
backup_dir = PROJECT / f"backup-before-auth-navigation-{timestamp}"
backup_dir.mkdir()

for name in PAGES:
    shutil.copy2(PROJECT / name, backup_dir / name)

# Changes write karein
for name, content in updated_pages.items():
    (PROJECT / name).write_text(content, encoding="utf-8")

AUTH_NAV.write_text(auth_nav_code.strip() + "\n", encoding="utf-8")

# Final verification
print("AUTHENTICATION NAVIGATION UPDATE COMPLETE")
print("Backup folder:", backup_dir.name)
print("New file: auth-nav.js")

for name in PAGES:
    content = (PROJECT / name).read_text(encoding="utf-8")
    print(
        name,
        "| Login:", 'id="login-nav-link"' in content,
        "| Profile:", 'id="profile-nav-link"' in content,
        "| Subscribe:", 'id="subscribe-nav-link"' in content,
        "| Explore Picks:", 'class="nav-cta"' in content,
        "| Auth JS:", "auth-nav.js" in content
    )
```

```python
from pathlib import Path
from datetime import datetime
import shutil
import subprocess

PROJECT = Path("/content/SmartPick-Hub")
JS_FILE = PROJECT / "auth-nav.js"

# Backup existing JS
backup = PROJECT / (
    "auth-nav-before-dropdown-" +
    datetime.now().strftime("%Y%m%d-%H%M%S") + ".js"
)

shutil.copy2(JS_FILE, backup)
print("JS BACKUP:", backup.name)

new_code = r'''// SmartPick Hub - Professional Authentication Navigation

const smartPickNavClient = window.supabase.createClient(
    SUPABASE_URL,
    SUPABASE_PUBLISHABLE_KEY
);

function updateSmartPickNavigation(session) {
    const loginLink = document.getElementById("login-nav-link");
    const profileMenu = document.getElementById("profile-menu");

    const isLoggedIn = Boolean(session);

    if (loginLink) {
        loginLink.hidden = isLoggedIn;
    }

    if (profileMenu) {
        profileMenu.hidden = !isLoggedIn;
    }

    if (isLoggedIn && session.user) {
        const email = session.user.email || "";
        const displayName =
            session.user.user_metadata?.full_name ||
            session.user.user_metadata?.name ||
            email.split("@")[0] ||
            "Account";

        const initial = displayName.trim().charAt(0).toUpperCase() || "U";

        const nameElement = document.getElementById("profile-name");
        const avatarElement = document.getElementById("profile-avatar");
        const dropdownName = document.getElementById("profile-dropdown-name");
        const dropdownEmail = document.getElementById("profile-dropdown-email");

        if (nameElement) nameElement.textContent = displayName;
        if (avatarElement) avatarElement.textContent = initial;
        if (dropdownName) dropdownName.textContent = displayName;
        if (dropdownEmail) dropdownEmail.textContent = email;
    }
}

function closeProfileDropdown() {
    const trigger = document.getElementById("profile-trigger");
    const dropdown = document.getElementById("profile-dropdown");

    if (trigger && dropdown) {
        trigger.setAttribute("aria-expanded", "false");
        dropdown.hidden = true;
    }
}

function initializeProfileDropdown() {
    const trigger = document.getElementById("profile-trigger");
    const dropdown = document.getElementById("profile-dropdown");
    const logoutButton = document.getElementById("logout-nav-button");
    const profileMenu = document.getElementById("profile-menu");

    if (trigger && dropdown) {
        trigger.addEventListener("click", () => {
            const isOpen = trigger.getAttribute("aria-expanded") === "true";

            trigger.setAttribute("aria-expanded", String(!isOpen));
            dropdown.hidden = isOpen;
        });

        document.addEventListener("click", (event) => {
            if (profileMenu && !profileMenu.contains(event.target)) {
                closeProfileDropdown();
            }
        });

        document.addEventListener("keydown", (event) => {
            if (event.key === "Escape") {
                closeProfileDropdown();
            }
        });
    }

    if (logoutButton) {
        logoutButton.addEventListener("click", async () => {
            logoutButton.disabled = true;
            logoutButton.textContent = "Logging out...";

            try {
                const { error } =
                    await smartPickNavClient.auth.signOut();

                if (error) throw error;

                window.location.replace("index.html");

            } catch (error) {
                console.error("Logout failed:", error.message);
                logoutButton.disabled = false;
                logoutButton.textContent = "Logout";
                alert("Logout failed. Please try again.");
            }
        });
    }
}

async function initializeSmartPickNavigation() {
    try {
        const { data, error } =
            await smartPickNavClient.auth.getSession();

        if (error) throw error;

        updateSmartPickNavigation(data.session);

    } catch (error) {
        console.error("Navigation session check failed:", error.message);
        updateSmartPickNavigation(null);
    }
}

smartPickNavClient.auth.onAuthStateChange(
    (event, session) => {
        updateSmartPickNavigation(session);
    }
);

initializeProfileDropdown();
initializeSmartPickNavigation();
'''

JS_FILE.write_text(new_code, encoding="utf-8")

result = subprocess.run(
    ["node", "--check", str(JS_FILE)],
    capture_output=True,
    text=True
)

print("\nJAVASCRIPT SYNTAX:",
      "PASS" if result.returncode == 0 else "ERROR")

if result.stderr:
    print(result.stderr)

print("Updated file:", JS_FILE)
```
